```python
drop_columns = ["Country","Gender","Diabetes_Risk" ,"Doctor_Consultation_Needed", "AI_Health_Recommendation","BMI_CATEGORY","Patient_ID"]
```

```python
status_dict = {'Low': 1, 'Moderate': 2, 'High': 3}
data['Physical_Activity_Level'] = data['Physical_Activity_Level'].map(status_dict)
data['Sugar_Intake_Level'] = data['Sugar_Intake_Level'].map(status_dict)
data['Stress_Level'] = data['Stress_Level'].map(status_dict)

```

```python
status_dict = {'Poor': 1, 'Average': 2, 'Healthy': 3}
data['Diet_Quality'] = data['Diet_Quality'].map(status_dict)

```

```python
status_dict = {'Former': 1, 'Current': 2, 'Never': 3}
data['Smoking_Status'] = data['Smoking_Status'].map(status_dict)
data['Alcohol_Consumption'] = data['Alcohol_Consumption'].map(status_dict)

```

```python
data
```

```python
status_dict = {'Frequently': 2, 'Occasionally': 1, 'Never': 3}
data['Alcohol_Consumption'] = data['Alcohol_Consumption'].map(status_dict)

```

```python
data.dtypes
```

```python
status_dict = {'Yes': 1, 'No': 0}
data['Family_History_Diabetes'] = data['Alcohol_Consumption'].map(status_dict)
data['Hypertension'] = data['Hypertension'].map(status_dict)
data['Heart_Disease'] = data['Heart_Disease'].map(status_dict)
data['Fatty_Liver'] = data['Fatty_Liver'].map(status_dict)
data['PCOS'] = data['PCOS'].map(status_dict)

```

```python
data.dtypes
```

```python
status_dict = {'Good': 2, 'Avg': 1,'Poor' : 0}
data['Medication_Adherence'] = data['Medication_Adherence'].map(status_dict)
status_dict = {'Urban' : 0 , 'Rural' :1}
data['Residence_Type'] = data['Residence_Type'].map(status_dict)



data['Work_Type'] = data['Work_Type'].astype('category')
```

```python
X = data.drop(["Diabetes_Risk_Score" , "Heart_Disease"], axis =1)
Y = data["Diabetes_Risk_Score"]
```

```python
from sklearn.model_selection import train_test_split

# X represents your features, y represents your target labels
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.23, random_state=42)
```

```python
!pip install xgboost
```

```python
import xgboost as xgb
model = xgb.XGBRegressor(
    objective='reg:squarederror', # Target continuous values
    n_estimators=100,             # Number of boosting rounds
    learning_rate=0.1,            # Step size shrinkage (eta)
    max_depth=5,                  # Maximum tree depth
    random_state=42
)




```

```python
model.fit(X_train , y_train)
```

```python
from sklearn.metrics import mean_squared_error

predictions = model.predict(X_test)
mse = mean_squared_error(y_test, predictions)
print(f"Mean Squared Error: {mse:.4f}")
```

```python
import pickle

# Save the trained model to a file
filename = 'model.pkl'
with open(filename, 'wb') as file:
    pickle.dump(model, file)
```
